```python
import pandas as pd

# 读取数据
df=pd.DataFrame({
   'student_id':[1,2,3,4,5],
    'score':[90,185,88,92,-87]
})

#计算均值和标准差
mean=df['score'].mean()
std=df['score'].std()

#计算每个数据点的z-score
df['z_score']=abs(df['score']-mean)/std

#检测异常值
outliers=df[df['z_score']>1]

outliers
```
